## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is a ranking task because the goal is to prioritize webpages based on their ranking signals. Instead of simply predicting whether a page is good or bad, the model ranks webpages from highest to lowest priority so that the pages most in need of attention are reviewed first.

This is a supervised learning task because historical search performance data provides labeled examples that the model can learn from. By using these historical patterns, the model can rank new webpages according to their likelihood of needing attention.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target of this task is to identify webpages that are likely to lose search visibility based on their ranking signals.

Since the actual benefit of refreshing a page cannot be measured in advance, historical search performance is used as a proxy. Features such as impressions, clicks, click-through rate (CTR), average position, and other ranking signals provide the information needed to estimate which pages should be prioritized for review and optimization.

The label comes from observed historical outcomes rather than a manually defined rule, making it suitable for supervised machine learning.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The success metric for my lane is Precision@50.

I chose this metric because the goal is to recommend the most important webpages first. If most of the top 50 pages suggested by the model actually need attention, then the model is performing well.

A good result would be a Precision@50 that is higher than a simple rule-based approach. This would show that the model is making more useful recommendations for deciding which pages should be reviewed and updated first.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule cannot accurately identify which webpages need attention because search performance depends on many factors working together. Metrics such as impressions, clicks, CTR, average position, content type, and search volume all influence a page's performance, and their relationships are not always simple.

Machine learning can learn these patterns from historical data instead of relying on one fixed rule. This makes the recommendations more flexible and accurate, especially when different pages behave differently over time.
