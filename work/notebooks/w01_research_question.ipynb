Research Question:

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

dataset_url = "https://raw.githubusercontent.com/ArshadNazir1/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"

data = pd.read_csv(dataset_url)

pages_with_lower_impressions = (
    data["impressions_last_30d"] < data["impressions_prev_30d"]
).sum()

percentage = (pages_with_lower_impressions / len(data)) * 100

print("Total content items:", len(data))
print("Content items with lower impressions:", pages_with_lower_impressions)
print("Percentage with lower impressions:", round(percentage, 2), "%")
```

The starter dataset contains 30,000 content items. Out of these, 19,716 content items received fewer impressions in the latest 30 days compared with the previous 30 days. This represents approximately 65.72% of the dataset. This observed decrease shows that there are many potential content-review candidates, which makes the selected lane worth investigating further. However, lower impressions alone do not mean that a page needs to be refreshed, because search performance may be affected by several other factors. Therefore, further analysis is needed before recommending any action.
